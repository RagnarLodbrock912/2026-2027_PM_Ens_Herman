```python
import numpy as np
import math
```

```python
A = np.array([
    [6, -1, 10, -1],
    [2, 1, 10, 7],
    [3, -2, -2, -1],
    [1, -12, 2, -1]
])

b = np.array([158, 128, 7, 17])

I = np.eye(4)
```

```python
det = np.linalg.det(A)
print(round(det))
```

```python
B = A.T @ A

eig = max(np.linalg.eigvals(B))
print(eig)
```

```python
C = I - (B / eig)
d = (A.T @ b) / eig
```

```python
alpha = max(np.linalg.eigvals(C))
print(alpha)
```

```python
x0 = [1, 2, 3, 4]
x1 = C @ x0 + d
```

```python
eps = 10**-4

rho2 = np.linalg.norm(x0 - x1, 2)

N_apr = int(
    np.log(eps * (1 - alpha) / rho2)
    / np.log(alpha)
) + 1

print(N_apr)
```

```python
err = (alpha / (1 - alpha)) * np.linalg.norm(x0 - x1, 2)
count = 1 
x_prev = x1

while err > eps:
    x_new = C @ x_prev + d
    err = (alpha / (1 - alpha)) * np.linalg.norm(x_prev - x_new, 2)
    x_prev = x_new
    count += 1

print(count)
print(x_new)
```
