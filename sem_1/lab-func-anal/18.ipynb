```python
import math
import matplotlib.pyplot as plt
import numpy as np
```

```python
eps = 10**-2
alpha = 1 / math.pi
a = -4 * math.pi
b = 4 * math.pi
```

```python
def solve(t, x0, i):
    if i == 0:
        return t + np.sin(x0(t) / np.pi)**2

    return t + np.sin(solve(t, x0, i - 1) / np.pi)**2
```

```python
def x0(t):
    return np.sin(t)
```

```python
t = np.linspace(a, b, 100000)

x1 = solve(t, x0, 0)

rho = np.max(np.abs(x0(t) - x1))

print(rho)
```

```python
N_apr = int(
    np.log(eps * (1 - alpha) / rho)
    / np.log(alpha)
) + 1

print(N_apr)
```

```python
x_final = solve(t, x0, N_apr)

plt.plot(t, x_final, label="x_final")
plt.plot(t, x0(t), label="x0")


plt.xlim(a, b)
plt.xlabel("t")
plt.ylabel("x(t)")
plt.legend()
plt.grid()
plt.show()
```
