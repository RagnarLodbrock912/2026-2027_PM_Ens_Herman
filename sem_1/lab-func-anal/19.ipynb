```python
import numpy as np
from scipy.integrate import quad
```

```python
alpha = 0.5
eps = 10**-5
lam = 1
```

```python
def x0(t):
    return 1 + 0*t


def next_x(x_prev):
    A = quad(lambda s: x_prev(s), 0, 1)[0]
    B = quad(lambda s: s * x_prev(s), 0, 1)[0]

    def x_new(t):
        return lam * (t * A - B) + (t * 0.5)

    return x_new
```

```python
t = np.linspace(0, 1, 100000)

x1 = next_x(x0)

rho = np.max(np.abs(x0(t) - x1(t)))

print(rho)
```

```python
N_apr = int(
    np.log(eps * (1 - alpha) / rho)
    / np.log(alpha)
) + 1

print(N_apr)
```

```python
x = x1

for i in range(N_apr - 1):
    x = next_x(x)
```

```python
a_coef = x(1) - x(0)
b_coef = x(0)

print(f"x(t) = {a_coef} * t + {b_coef}")
```

```python
def x_exact(t):
    return (9 / 13) * t - (2 /13)

err = np.max(np.abs(x(t) - x_exact(t)))
print(err)
```
